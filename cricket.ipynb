```python
df["Boundary Runs"] = df["4s"]*4 + df["6s"]*6
```

```python
df["Singles"] = df["runs"] - df["Boundary Runs"]
```

```python
df["Boundary %"] = (df["Boundary Runs"]/df["runs"])*100
```

```python
df["Boundary %"] = (df["Boundary Runs"]/df["runs"].replace(0,1))*100
```

```python
df["Fifty"] = df["runs"]>=50
```

```python
df["Century"] = df["runs"]>=100
```
